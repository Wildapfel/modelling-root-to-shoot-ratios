```python
params_gamma = pd.read_csv(PATH_PARAMS, index_col=0)
params_gamma.head()
```

```python
beta = Beta()
```

```python
a_beta = params_gamma.loc["sprout_a"]
b_beta = params_gamma.loc["root_a"]
x, y = beta.PDF(a=a_beta, b=b_beta, xmin=0, xmax=1.0)
```

```python
fig = curve_fit(df=df,
                column="Ratio",
                title="beta fit ratio",
                curves=[[x,y]],
                labels=["beta"])
fig.savefig("../reports/figures/ratio_theoretical_beta_fit.png", bbox_inches="tight")
```
